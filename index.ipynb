```python
def Resumo_noticia(text):
  template = """
  Resuma essa noticia em um paragrafo e seja objetivo: {text}
  """

  prompt = PromptTemplate.from_template(template=template)
  chat = ChatGroq(api_key=api_key, model='llama-3.1-8b-instant')
  chain = prompt | chat
  
  print(chain.invoke(text).content)
```

```python
def Conteudo_noticia(url_noticia):
  response = requests.get(f"https://www.binance.com/{url_noticia}")
  soup = BeautifulSoup(response.content, "html.parser")
  for title in soup.find_all('p'):
    Resumo_noticia(title.text)
```
